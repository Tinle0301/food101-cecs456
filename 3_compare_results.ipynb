# Comparison: scratch CNN vs. ResNet-50
Run after both model notebooks. Reads the saved results in `results/`.

```python
import json
import matplotlib.pyplot as plt
from common import RESULTS_DIR

names = {'scratch': 'Scratch CNN', 'transfer': 'ResNet-50'}
metrics = {m: json.loads((RESULTS_DIR / m / 'test_metrics.json').read_text()) for m in names}
history = {m: json.loads((RESULTS_DIR / m / 'history.json').read_text()) for m in names}
```

**Table 1.** Test set metrics (precision, recall and F1 are macro-averaged over the 20 classes)

```python
print(f"{'':12s} {'Loss':>8s} {'Accuracy':>9s} {'Precision':>10s} {'Recall':>8s} {'F1':>8s} {'Minutes':>8s}")
for m, n in names.items():
    r = metrics[m]
    print(f"{n:12s} {r['loss']:8.4f} {r['accuracy']:9.4f} {r['precision']:10.4f} {r['recall']:8.4f} {r['f1']:8.4f} {history[m]['train_minutes']:8}")
```

```python
fig, ax = plt.subplots(1, 2, figsize=(11, 4))
for m, n in names.items():
    h = history[m]
    ep = range(1, len(h['val_acc']) + 1)
    ax[0].plot(ep, h['val_loss'], label=n)
    ax[1].plot(ep, h['val_acc'], label=n)
ax[0].set(title='Validation loss', xlabel='Epoch', ylabel='Loss')
ax[1].set(title='Validation accuracy', xlabel='Epoch', ylabel='Accuracy')
for a in ax:
    a.legend()
fig.tight_layout()
```

```python
for m, n in names.items():
    print(n, 'most confused (actual -> predicted):')
    for actual, pred, count in metrics[m]['top_confused'][:5]:
        print(f'   {actual} -> {pred}: {count}')
```
